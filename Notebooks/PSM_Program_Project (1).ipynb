```python
import pandas as pd
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
X=[['hujan','pdd','mdpl2','defor']]
```

```python
pdd_vif = pdd.copy()
pdd_vif['Kabupaten'] = pdd_vif['Kabupaten dan Kota'].astype(str).apply(lambda x: x.split(' ', 2)[2] if ' ' in x and len(x.split(' ')) > 2 and x.startswith('33') else x)
pdd_vif = pdd_vif.rename(columns={'Unnamed: 4': 'Kepadatan_Penduduk'})
pdd_vif = pdd_vif[['Kabupaten', 'Kepadatan_Penduduk']]

mdpl2_vif = mdpl2.copy()
mdpl2_vif['Kabupaten'] = mdpl2_vif['Kabupaten dan Kota'].astype(str).apply(lambda x: x.split(' ', 2)[2] if ' ' in x and len(x.split(' ')) > 2 and x.startswith('33') else x)
mdpl2_vif = mdpl2_vif.rename(columns={'Unnamed: 1': 'Ketinggian_Wilayah'})
mdpl2_vif = mdpl2_vif[['Kabupaten', 'Ketinggian_Wilayah']]

defor_vif = defor.copy()
defor_vif['Kabupaten'] = defor_vif['Kabupaten / Kota'].astype(str).apply(lambda x: x.split(' ', 2)[2] if ' ' in x and len(x.split(' ')) > 2 and x.startswith('33') else x)
defor_vif = defor_vif.rename(columns={'Unnamed: 6': 'Total_Deforestasi'})
defor_vif = defor_vif[['Kabupaten', 'Total_Deforestasi']]

hujan_vif = hujan.rename(columns={'kabupaten': 'Kabupaten', 'curah;': 'curah_hujan'})
hujan_vif = hujan_vif[['Kabupaten', 'curah_hujan']]

merged_data = hujan_vif.merge(pdd_vif, on='Kabupaten', how='inner')
merged_data = merged_data.merge(mdpl2_vif, on='Kabupaten', how='inner')
merged_data = merged_data.merge(defor_vif, on='Kabupaten', how='inner')

X_for_vif = merged_data.drop('Kabupaten', axis=1)

X_for_vif = X_for_vif.apply(pd.to_numeric, errors='coerce')

X_for_vif = X_for_vif.dropna()

for col in X_for_vif.columns:
    if X_for_vif[col].std() == 0:
        X_for_vif = X_for_vif.drop(columns=[col])

if X_for_vif.empty or X_for_vif.shape[1] < 2:
    print("Not enough valid numerical columns for VIF calculation after cleaning.")
else:
    vif = pd.DataFrame()
    vif["variabel"] = X_for_vif.columns
    vif["VIF"] = [variance_inflation_factor(X_for_vif.values, i) for i in range(X_for_vif.shape[1])]

    print(vif)
```

```python
import statsmodels.api as sm
import pandas as pd

def clean_kabupaten_name_for_merge(name_series):
    cleaned_series = name_series.astype(str)
    cleaned_series = cleaned_series.apply(lambda x: ' '.join(x.split(' ')[1:]) if x.split(' ')[0].isdigit() else x)
    cleaned_series = cleaned_series.str.replace('Kabupaten ', '', regex=False)
    cleaned_series = cleaned_series.str.replace('Kota ', '', regex=False)
    return cleaned_series.str.strip()

hujan_cleaned = hujan.rename(columns={'kabupaten': 'Kabupaten', 'curah;': 'curah_hujan'}).copy()
hujan_cleaned['Kabupaten'] = clean_kabupaten_name_for_merge(hujan_cleaned['Kabupaten'])
hujan_cleaned = hujan_cleaned[['Kabupaten', 'curah_hujan']]

pdd_cleaned = pdd.copy()
pdd_cleaned['Kabupaten'] = clean_kabupaten_name_for_merge(pdd_cleaned['Kabupaten dan Kota'])
pdd_cleaned = pdd_cleaned.rename(columns={'Unnamed: 4': 'Kepadatan_Penduduk'})
pdd_cleaned = pdd_cleaned[['Kabupaten', 'Kepadatan_Penduduk']]

mdpl2_cleaned = mdpl2.copy()
mdpl2_cleaned['Kabupaten'] = clean_kabupaten_name_for_merge(mdpl2_cleaned['Kabupaten dan Kota'])
mdpl2_cleaned = mdpl2_cleaned.rename(columns={'Unnamed: 1': 'Ketinggian_Wilayah'})
mdpl2_cleaned = mdpl2_cleaned[['Kabupaten', 'Ketinggian_Wilayah']]

defor_cleaned = defor.copy()
defor_cleaned['Kabupaten'] = clean_kabupaten_name_for_merge(defor_cleaned['Kabupaten / Kota'])
defor_cleaned = defor_cleaned.rename(columns={'Unnamed: 6': 'Total_Deforestasi'})
defor_cleaned = defor_cleaned[['Kabupaten', 'Total_Deforestasi']]

bencana_cleaned = bencana.copy()
bencana_cleaned['Kabupaten'] = clean_kabupaten_name_for_merge(bencana_cleaned['Kabupaten/Kota'])
bencana_cleaned = bencana_cleaned[['Kabupaten', 'Jumlah Bencana Alam - Banjir']]

merged_data_for_model = hujan_cleaned.merge(pdd_cleaned, on='Kabupaten', how='inner')
merged_data_for_model = merged_data_for_model.merge(mdpl2_cleaned, on='Kabupaten', how='inner')
merged_data_for_model = merged_data_for_model.merge(defor_cleaned, on='Kabupaten', how='inner')
merged_data_for_model = merged_data_for_model.merge(bencana_cleaned, on='Kabupaten', how='inner')

y = merged_data_for_model['Jumlah Bencana Alam - Banjir']
X = merged_data_for_model.drop(['Kabupaten', 'Jumlah Bencana Alam - Banjir'], axis=1)

X = X.apply(pd.to_numeric, errors='coerce').dropna()
y = y.apply(pd.to_numeric, errors='coerce').dropna()

common_index = X.index.intersection(y.index)
X = X.loc[common_index]
y = y.loc[common_index]

X = sm.add_constant(X)

model = sm.GLM(y, X, family=sm.families.Poisson())
hasil = model.fit()

print(hasil.summary())
```

```python
prediksi = hasil.predict(X)
print(prediksi)
```

```python
merged_data_for_model['prediksi'] = prediksi
print(merged_data_for_model)
```
